## 4. Preprocessing and Feature Engineering

This section prepares the dataset for predictive modeling. The model intentionally uses article text only as input, so that engineered fields such as `sentiment_score` do not leak the answer into the features.

```python
df_model = df.copy()
df_model = df_model.dropna(subset=['content_clean', 'sentiment_label'])
df_model = df_model[df_model['content_clean'].str.strip().ne('')]
df_model = df_model[df_model['sentiment_label'].isin(['Positive', 'Negative', 'Neutral'])]

df_model['computed_word_count'] = df_model['content_clean'].str.split().str.len()
df_model['pub_year'] = df_model['date'].dt.year
df_model['pub_month'] = df_model['date'].dt.month

features_summary = pd.DataFrame({
    'feature': ['content_clean', 'computed_word_count', 'pub_year', 'pub_month', 'publication', 'sentiment_label'],
    'purpose': [
        'Primary text input for classification and retrieval',
        'Article length and information density indicator',
        'Temporal segmentation for business analysis',
        'Monthly trend analysis',
        'Source-level analysis and possible bias monitoring',
        'Target label for implemented sentiment-routing model',
    ]
})
display(features_summary)
print('Modeling shape:', df_model.shape)
display(df_model['sentiment_label'].value_counts())
```

## 5. Train-Test Split

A stratified 80/20 split keeps Positive, Negative, and Neutral classes represented in both training and test data.

```python
X = df_model['content_clean'].astype(str)
y = df_model['sentiment_label'].astype(str)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y,
)

split_table = pd.DataFrame({
    'full_dataset': y.value_counts(),
    'train_set': y_train.value_counts(),
    'test_set': y_test.value_counts(),
}).fillna(0).astype(int)

display(split_table)
print('Train size:', len(X_train))
print('Test size:', len(X_test))
```

## 6. Baseline and Advanced Models

Three models are compared:

1. **Majority benchmark:** predicts the largest class for every article.
2. **TF-IDF + Multinomial Naive Bayes:** interpretable traditional NLP baseline.
3. **TF-IDF + Dense Neural Network:** advanced model using a small MLP classifier over TF-IDF vectors.

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    min_df=2,
    max_features=900,
)

models = {
    'Majority benchmark': Pipeline([
        ('tfidf', vectorizer),
        ('model', DummyClassifier(strategy='most_frequent')),
    ]),
    'TF-IDF + Naive Bayes': Pipeline([
        ('tfidf', TfidfVectorizer(lowercase=True, stop_words='english', min_df=2, max_features=900)),
        ('model', MultinomialNB(alpha=1.0)),
    ]),
    'TF-IDF + DNN': Pipeline([
        ('tfidf', TfidfVectorizer(lowercase=True, stop_words='english', min_df=2, max_features=900)),
        ('model', MLPClassifier(
            hidden_layer_sizes=(36,),
            activation='relu',
            solver='adam',
            learning_rate_init=0.01,
            max_iter=45,
            random_state=RANDOM_STATE,
        )),
    ]),
}

print('Models ready:', list(models.keys()))
```

```python
def evaluate_model(name, model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    precision, recall, f1, support = precision_recall_fscore_support(
        y_test, pred, labels=['Positive', 'Negative', 'Neutral'], zero_division=0
    )
    result = {
        'model': name,
        'accuracy': accuracy_score(y_test, pred),
        'macro_f1': f1_score(y_test, pred, average='macro', zero_division=0),
        'weighted_f1': f1_score(y_test, pred, average='weighted', zero_division=0),
        'prediction': pred,
        'per_class': pd.DataFrame({
            'class': ['Positive', 'Negative', 'Neutral'],
            'precision': precision,
            'recall': recall,
            'f1': f1,
            'support': support,
        }),
    }
    return result

results = []
trained_models = {}

for name, model in models.items():
    print('\nTraining:', name)
    result = evaluate_model(name, model, X_train, X_test, y_train, y_test)
    results.append(result)
    trained_models[name] = model
    print(f"Accuracy: {result['accuracy']:.4f} | Macro F1: {result['macro_f1']:.4f} | Weighted F1: {result['weighted_f1']:.4f}")
```

## 7. Evaluation Results

```python
metrics_df = pd.DataFrame([
    {
        'Model': r['model'],
        'Accuracy': r['accuracy'],
        'Macro F1': r['macro_f1'],
        'Weighted F1': r['weighted_f1'],
    }
    for r in results
])

display(metrics_df.style.format({'Accuracy': '{:.2%}', 'Macro F1': '{:.2%}', 'Weighted F1': '{:.2%}'}))

fig, ax = plt.subplots(figsize=(9, 4))
x = np.arange(len(metrics_df))
width = 0.35
ax.bar(x - width/2, metrics_df['Accuracy'], width, label='Accuracy', color='#4C78A8')
ax.bar(x + width/2, metrics_df['Macro F1'], width, label='Macro F1', color='#F58518')
ax.set_xticks(x)
ax.set_xticklabels(metrics_df['Model'], rotation=15, ha='right')
ax.set_ylim(0, 1.08)
ax.set_title('Model Comparison')
ax.set_ylabel('Score')
ax.legend()
save_current_fig('final_model_comparison.png')
plt.show()
```

```python
for r in results:
    print('\n' + '=' * 80)
    print(r['model'])
    print('=' * 80)
    display(r['per_class'].style.format({'precision': '{:.2%}', 'recall': '{:.2%}', 'f1': '{:.2%}'}))
    print(classification_report(y_test, r['prediction'], zero_division=0))
```

```python
best_model_name = metrics_df.sort_values(['Macro F1', 'Accuracy'], ascending=False).iloc[0]['Model']
best_prediction = next(r['prediction'] for r in results if r['model'] == best_model_name)

fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay.from_predictions(
    y_test,
    best_prediction,
    labels=['Positive', 'Negative', 'Neutral'],
    cmap='Blues',
    ax=ax,
    colorbar=False,
)
ax.set_title(f'Confusion Matrix - {best_model_name}')
save_current_fig('final_confusion_matrix.png')
plt.show()
```

## 9. Lightweight BI Assistant Prototype

The full project idea is an LLM-based BI assistant. Without calling an external LLM API, the notebook can still demonstrate the assistant workflow using TF-IDF retrieval plus structured summaries. This section retrieves articles relevant to a business question and returns sentiment counts, source counts, and representative article titles.

```python
retrieval_vectorizer = TfidfVectorizer(lowercase=True, stop_words='english', min_df=1, max_features=2000)
article_matrix = retrieval_vectorizer.fit_transform(df_model['content_clean'])

def business_query_assistant(query, top_k=8):
    query_vec = retrieval_vectorizer.transform([query])
    scores = (article_matrix @ query_vec.T).toarray().ravel()
    top_idx = scores.argsort()[::-1][:top_k]
    retrieved = df_model.iloc[top_idx].copy()
    retrieved['relevance_score'] = scores[top_idx]
    retrieved = retrieved[retrieved['relevance_score'] > 0]
    
    if retrieved.empty:
        return {
            'query': query,
            'summary': 'No relevant articles found for this query.',
            'articles': pd.DataFrame(),
        }
    
    sentiment_counts = retrieved['sentiment_label'].value_counts().to_dict()
    source_counts = retrieved['publication'].value_counts().to_dict()
    top_titles = retrieved[['title', 'publication', 'sentiment_label', 'relevance_score']].drop_duplicates('title')
    
    summary = (
        f"For the query '{query}', the assistant retrieved {len(retrieved)} relevant article rows. "
        f"Sentiment mix: {sentiment_counts}. Main sources: {source_counts}. "
        "A manager can review the listed titles first because they are the highest-scoring matches."
    )
    return {
        'query': query,
        'summary': summary,
        'articles': top_titles,
    }

demo = business_query_assistant('interest rates inflation federal reserve risk', top_k=10)
print(demo['summary'])
display(demo['articles'])
```

```python
demo_queries = [
    'technology company revenue growth opportunity',
    'oil prices production agreement risk',
    'unemployment rate economy labor market',
    'bitcoin institutional investors market',
]

for q in demo_queries:
    demo = business_query_assistant(q, top_k=5)
    print('\n' + '-' * 90)
    print(demo['summary'])
    display(demo['articles'])
```
